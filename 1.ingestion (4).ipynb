```python
import requests
import os
import pandas as pd

# 1. SETUP PATHS (Adjust if your folder is named differently)
BASE_PATH = "/home/jovyan/labs/project_work"
STREAM_PATH = f"{BASE_PATH}/gdelt_stream"

if not os.path.exists(STREAM_PATH):
    os.makedirs(STREAM_PATH)
    print(f"📁 Created folder: {STREAM_PATH}")

# 2. CONFLICT COUNTRIES (The filter)
CONFLICT_COUNTRIES = {
    'UK':'Ukraine', 
    'RS':'Russia', 
    'IS':'Israel', 
    'GZ':'Gaza', 
    'US':'USA', 
    'IN':'India'
}

# 3. FETCH THE LATEST FILENAME
print("📡 Checking GDELT for the latest file...")
r = requests.get("http://data.gdeltproject.org/gdeltv2/lastupdate.txt")
url = r.text.splitlines()[0].split()[-1] 

# We add 'FORCE_' to the name so Spark sees it as a brand new event
file_name = "FORCE_" + url.split("/")[-1].replace(".zip", ".csv")
save_path = f"{STREAM_PATH}/{file_name}"

# 4. DOWNLOAD AND FILTER
print(f"📥 Downloading: {url}")
try:
    df = pd.read_csv(url, sep='\t', header=None, compression='zip', on_bad_lines='skip')
    
    # Filter for the countries in your list (Column 54 is Actor1CountryCode)
    df_filtered = df[df[54].isin(CONFLICT_COUNTRIES.keys())].copy()

    if len(df_filtered) > 0:
        # SAVE TO THE STREAM FOLDER
        df_filtered.to_csv(save_path, sep='\t', index=False, header=False)
        print("-" * 30)
        print(f"✅ SUCCESS!")
        print(f"📄 File Saved: {file_name}")
        print(f"📊 Rows Captured: {len(df_filtered)}")
        print("-" * 30)
    else:
        print("⚠️ Downloaded successfully, but no matching countries found in this 15-min window.")

except Exception as e:
    print(f"❌ Error during download: {e}")
```
